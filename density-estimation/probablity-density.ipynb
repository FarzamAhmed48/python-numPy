```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.random import normal 
```

## Sample Data

```python
sample = normal(loc=50,scale=5,size=1000)

#mean is 50 and standard deviation is 5 
sample.mean()
```

## Plotting Sample in matplotlib histogram

```python
plt.hist(sample,bins=10)
```

### Calculating Sample Mean And Standard Deviation

```python
sample_mean = sample.mean()
sample_std=sample.std()
```

```python
#fitting distribution with the above calculated parameters 
from scipy.stats import norm
dist=  norm(sample_mean,sample_std)
```

```python
values= np.linspace(sample.min(),sample.max(),100)
```

```python
sample.max()
```

```python
probablities = [dist.pdf(value) for value in values]
```

```python
plt.hist(sample,bins=10,density=True)
plt.plot(values,probablities)
```
