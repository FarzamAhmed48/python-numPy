## Advance Operations with Business Example

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [restaurant_id,2021,2022,2023,2024]

sales_data=np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryanni
    [2, 120000, 140000, 160000, 190000], # Beijing Bites
    [3, 200000, 230000, 260000, 300000], # Pizza Hub
    [4, 180000, 210000, 240000, 270000], # Burger Point
    [5, 160000, 185000, 205000, 230000], # Chai Point
])

print("=========FoodPanda sales Analysis=========")
print(f"The shape of the current data is {sales_data.shape}")
print(f"Selecting the data till the 3rd row:\n {sales_data[:3]}")
print(f"Selecting all the data except the restaturant ids:\n {sales_data[:,1:]}")
print(f"Selecing the 2021 sales data of all restaurants:\n {sales_data[:,2]}")
```

```python
id_less_data=sales_data[:,1:]
print(id_less_data)
print(f"Sum of sales of the restaurants of all years {np.sum(id_less_data,axis=1)}")


print(f"Minimum sales of the restaurants year wise {np.min(id_less_data,axis=1)}")
```

```python
print(f"Maximum sales yearly {np.max(id_less_data,axis=0)}")

average_sales_restaurant = np.mean(id_less_data,axis=1)
print("Average sales per restaurant",average_sales_restaurant)

cumulative_sales_restaurant=np.cumsum(id_less_data,axis=1)
print("Cumulative sales ", cumulative_sales_restaurant)
```

## MatplotLib Basic

```python
plt.figure(figsize=(5,5))
plt.plot(np.mean(cumulative_sales_restaurant,axis=0))
plt.title("Average Cumulative Sales Per Year Accross all restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
