```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 10  #Number of trials
p = 0.5 #probablity of success
size = 1000 #Number of samples to generate

binomial_dist = np.random.binomial(n,p,size)
plt.hist(binomial_dist,density=True)
plt.show()
```

## Sample Means Graph

```python
import numpy as np
import matplotlib.pyplot as plt

num_samples = 1000
sample_size = 30
distribution_range = (0,1)

sample = np.random.uniform(distribution_range[0],distribution_range[1],(num_samples,sample_size))

sample_means=np.mean(sample,axis=0)

plt.hist(sample_means,density=True,bins=30,edgecolor = 'black')
plt.title("Histogram of Sample Means")
plt.xlabel("Sample Mean")
plt.ylabel("Density")

plt.show()
```
