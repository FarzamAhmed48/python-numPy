##Numpy array and basics

```python
import numpy as np
```

## Creating Numpy arrays from list

```python
arr_1d=  np.array([1,2,3,4,5])
print(f"This is 1d array {arr_1d}")

arr_2d=np.array([[1,2,3],[4,5,6]])
print(f"This is 2d array{arr_2d}")
```

## List vs Numpy array

```python
py_list = [1,2,3]
print(f"List multiplication {py_list *2}")

np_arr = np.array([1,2,3])
print(f"Numpy array multiplication {np_arr *2}")

import time

start=time.time()
li=[i*2 for i in range(1000000)]
print(f"Calculation time for list mul is: {time.time() - start}")

startnp=time.time()
arr=np.arange(1000000)*2
print(f"Calculation time for arr mul is: {time.time() - startnp}")
```

## Creating Arrays in numpy from scratch

```python
zeros_arr  = np.zeros((3,3))
print(f"The array of zeros \n {zeros_arr}")

ones_arr  = np.ones((3,3))
print(f"The array of ones \n {ones_arr}")

constant_arr = np.full((3,3),7)
print(f"The array of contant \n {constant_arr}")

random_arr = np.random.random((3,4))
print(f"The random array from numpy is \n{random_arr}")
```
