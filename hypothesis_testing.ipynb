## Paired T-Tests

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

before=np.array([80,92,75,68,65,78,73,90,70,88,76,84,82,77,91])
after=np.array([78,93,81,67,88,76,74,91,69,88,77,81,80,79,88])

```

```python
difference = before-after
```

```python
plt.hist(difference)
plt.title("Histogram of Weight Difference")
plt.xlabel("Weight differrences (Kg)")
plt.ylabel("Frequency")
plt.show()

shipro_test=  stats.shapiro(difference)
print("Shapiro-Wilk Test", shipro_test)
```

```python
mean_diff = np.mean(difference)
std_diff  = np.std(difference,ddof=1)
```

```python
n=len(difference)
t_statistic=  mean_diff/(std_diff/std_diff/np.sqrt(n))
df=n-1
```

```python
t_statistic
```

```python
alpha=0.05
p_value = 1-stats.t.cdf(t_statistic,df)
```

```python
p_value
```
